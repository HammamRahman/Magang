```python
import folium

def create_map(df):
  i = 0
  m = folium.Map(location=[df['Latitude'][i], df['Longitude'][i]], zoom_start=12)

  for i, row in df.iterrows():
    folium.Marker(
        location=[row['Latitude'], row['Longitude']],
            popup=f"Nama Usaha: {row['Nama Usaha']}<br>\n"
                  f"Jenis Usaha: {row['Jenis Usaha']}<br>\n"
                  f"Alamat: {row['Alamat']}<br>\n"
                  f"Koordinat: {row['Latitude']}, {row['Longitude']}<br>\n"
        ).add_to(m)
  i =+ 1
  return m

m = create_map(df)

m
```

```python
df_potensi = df[df['Status'].str.lower() == 'potensi']
df_pelanggan = df[df['Status'].str.lower() == 'pelanggan']
def create_map(df_potensi, df_pelanggan):
  i = 0
  map = folium.Map(location=[df['Latitude'][i], df['Longitude'][i]], zoom_start=12)

  # Menambahkan marker untuk potensi
  for i, row in df_potensi.iterrows():
    folium.Marker(
      location=[row['Latitude'], row['Longitude']],
        popup=f"Nama Usaha: {row['Nama Usaha']}<br>\n"
              f"Jenis Usaha: {row['Jenis Usaha']}<br>\n"
              f"Alamat: {row['Alamat']}<br>\n"
              f"Koordinat: {row['Latitude']}, {row['Longitude']}<br>\n",
        icon=folium.Icon(color='blue', icon='info-sign')
   ).add_to(map)

    # Menambahkan marker untuk pelanggan
  for i, row in df_pelanggan.iterrows():
    folium.Marker(
      location=[row['Latitude'], row['Longitude']],
        popup=f"Nama Usaha: {row['Nama Usaha']}<br>\n"
              f"Jenis Usaha: {row['Jenis Usaha']}<br>\n"
              f"Alamat: {row['Alamat']}<br>\n"
              f"Koordinat: {row['Latitude']}, {row['Longitude']}<br>\n",
      icon=folium.Icon(color='green', icon='ok-sign')
    ).add_to(map)
  i =+ 1
  return map

map = create_map(df_potensi, df_pelanggan)

map
```
